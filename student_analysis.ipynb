This time, Task 3 will be implemented using your actual StudentsPerformance.csv dataset containing 1,000 student records.

## **3.3 Logic and Planning**

2. Create a function to convert marks into grades using if-elif-else statements.
3. Create a function to determine whether a student has Passed or Failed.
4. Create a function to calculate the average of a list of marks.
5. Apply these functions to all 1,000 student records.
6. Calculate individual subject grades and each student's overall average.
7. Test the functions using boundary values: 0, 39, 40, 74, 75, 89, 90 and 100.
8. Check invalid marks such as -5 and 105.

## **3.4 Explanation**

- **Grade Function:** Assigns a grade according to the marks obtained by a student.
- **Pass/Fail Function:** Returns Failed if any subject mark is below 40; otherwise, returns Passed.
- **Average Function:** Calculates the average of Mathematics, Reading and Writing marks.
- **Dataset Application:** Applies the functions to all student records instead of testing only manually entered marks.

## **Function Summary**

| **Function** | **Purpose** |
|---|---|
| `get_grade()` | Converts marks into a grade. |
| `check_result()` | Determines whether a student has Passed or Failed. |
| `calculate_average()` | Calculates the average marks. |

## **3.5 Logic Challenge**

**What should a function do for invalid marks such as -5 or 105?**

The function should reject invalid marks because valid examination marks must be between 0 and 100. A `ValueError` will be raised when marks are outside this range.

This prevents incorrect values from affecting the final results.

```python
# TASK 3: BUILD THE GRADING RULES

import pandas as pd

# 1. Read the dataset
df = pd.read_csv("/content/StudentsPerformance.csv")

# 2. Function to calculate grade
def get_grade(marks):

    if marks < 0 or marks > 100:
        raise ValueError("Invalid marks")

    if marks >= 90:
        return "A"
    elif marks >= 75:
        return "B"
    elif marks >= 60:
        return "C"
    elif marks >= 40:
        return "D"
    else:
        return "F"


# 3. Function to check Pass or Fail
def check_result(marks):

    for mark in marks:

        if mark < 0 or mark > 100:
            raise ValueError("Invalid marks")

        if mark < 40:
            return "Failed"

    return "Passed"


# 4. Function to calculate average
def calculate_average(marks):

    if len(marks) == 0:
        raise ValueError("Marks list cannot be empty")

    for mark in marks:

        if mark < 0 or mark > 100:
            raise ValueError("Invalid marks")

    return sum(marks) / len(marks)


# 5. Apply grade function to each subject
df["Math Grade"] = df["math score"].apply(get_grade)
df["Reading Grade"] = df["reading score"].apply(get_grade)
df["Writing Grade"] = df["writing score"].apply(get_grade)


# 6. Calculate overall average
df["Average"] = df.apply(
    lambda row: calculate_average([
        row["math score"],
        row["reading score"],
        row["writing score"]
    ]),
    axis=1
)


# 7. Calculate overall grade
df["Overall Grade"] = df["Average"].apply(get_grade)


# 8. Calculate Pass or Fail
df["Result"] = df.apply(
    lambda row: check_result([
        row["math score"],
        row["reading score"],
        row["writing score"]
    ]),
    axis=1
)


# 9. Display the results
print("STUDENT GRADING RESULTS")
display(df[[
    "math score", "Math Grade",
    "reading score", "Reading Grade",
    "writing score", "Writing Grade",
    "Average", "Overall Grade", "Result"
]].head(10))


# 10. Test boundary values
print("\nBOUNDARY VALUE TESTING")

test_marks = [0, 39, 40, 74, 75, 89, 90, 100]

for mark in test_marks:
    print(mark, "=", get_grade(mark))


# 11. Test invalid marks
print("\nINVALID MARKS TESTING")

for mark in [-5, 105]:

    try:
        print(get_grade(mark))

    except ValueError as e:
        print(mark, ":", e)
```

## **4.5 Logic Challenge**

**Q1. How can the second-highest scorer be found without using sorting?**

The programme maintains two variables: the highest percentage and the second-highest distinct percentage. It compares each student's percentage with these values and updates them whenever a higher score is found. This identifies the second-highest percentage without sorting the records.

**Q2. How should the programme handle two toppers with the same percentage?**

If two or more students achieve the same highest percentage, all of them should be recognised as toppers. The programme should display all students who share the highest percentage. The second-highest scorer should be identified using the next distinct percentage.

## **4.6 Dataset Note**

The programme uses the first 10 actual records from the StudentsPerformance dataset. The dataset contains Mathematics, Reading and Writing marks but does not contain student names or roll numbers. Therefore, generated labels such as Student 1, Student 2 and Student 3 are used only for identification in the report cards.

```python
# TASK 4: SOLVE IT WITH PLAIN PYTHON

import csv

# 1. Read the first 10 records from the dataset
students = []

with open("/content/StudentsPerformance.csv", "r", encoding="utf-8-sig") as file:
    reader = csv.DictReader(file)

    for i, row in enumerate(reader):

        if i == 10:
            break

        student = {
            "name": "Student " + str(i + 1),
            "math": int(row["math score"]),
            "reading": int(row["reading score"]),
            "writing": int(row["writing score"])
        }

        students.append(student)


# 2. Initialise grade count
grade_count = {
    "A": 0,
    "B": 0,
    "C": 0,
    "D": 0,
    "F": 0
}


# 3. Calculate total, percentage, grade and result
for student in students:

    total = 0
    failed_subjects = []

    for subject in ["math", "reading", "writing"]:

        marks = student[subject]
        total = total + marks

        if marks < 40:
            failed_subjects.append(subject)

    percentage = total / 3

    # Assign grade
    if percentage >= 90:
        grade = "A"
    elif percentage >= 75:
        grade = "B"
    elif percentage >= 60:
        grade = "C"
    elif percentage >= 40:
        grade = "D"
    else:
        grade = "F"

    # Determine result
    if len(failed_subjects) > 0:
        result = "Failed"
    else:
        result = "Passed"

    student["total"] = total
    student["percentage"] = percentage
    student["grade"] = grade
    student["result"] = result
    student["failed_subjects"] = failed_subjects

    grade_count[grade] = grade_count[grade] + 1


# 4. Print report cards
print("STUDENT REPORT CARDS")
print("=" * 45)

for student in students:

    print("\nStudent:", student["name"])
    print("Mathematics:", student["math"])
    print("Reading:", student["reading"])
    print("Writing:", student["writing"])
    print("Total:", student["total"], "/ 300")
    print("Percentage:", round(student["percentage"], 2), "%")
    print("Grade:", student["grade"])
    print("Result:", student["result"])

    print("-" * 45)


# 5. Find topper and second-highest percentage
highest = -1
second_highest = -1

for student in students:

    percentage = student["percentage"]

    if percentage > highest:
        second_highest = highest
        highest = percentage

    elif percentage < highest and percentage > second_highest:
        second_highest = percentage


# 6. Display all toppers
print("\nTOPPER")
print("=" * 45)

for student in students:

    if student["percentage"] == highest:
        print(student["name"], "-", round(highest, 2), "%")


# 7. Display second-highest scorer(s)
print("\nSECOND-HIGHEST SCORER")
print("=" * 45)

if second_highest == -1:
    print("No second-highest score available.")

else:
    for student in students:

        if student["percentage"] == second_highest:
            print(student["name"], "-", round(second_highest, 2), "%")


# 8. Display grade count
print("\nGRADE COUNT")
print("=" * 45)

for grade in grade_count:
    print("Grade", grade, ":", grade_count[grade])


# 9. Display failed students and failed subjects
print("\nFAILED STUDENTS")
print("=" * 45)

for student in students:

    if student["result"] == "Failed":

        print(student["name"])
        print("Failed Subjects:", student["failed_subjects"])
```

The original array contains 10 students and 3 subjects.

Shape: (10, 3)

| Operation | Axis | Expected Shape |
|---|---|---|
| Subject-wise average | 0 | (3,) |
| Student-wise total | 1 | (10,) |
| Subject-wise standard deviation | 0 | (3,) |
| Count below 40 per subject | 0 | (3,) |

- axis=0 performs calculations subject-wise.
- axis=1 performs calculations student-wise.

## 5.5 Grace Marks Calculation

Five grace marks will be added to Mathematics marks. However, no mark should exceed 100.

For example, if a student scores 96, the final mark after adding grace marks will be 100 instead of 101.

## 5.6 Logic Challenge

1. Changing the axis changes the direction of calculation.
2. The number of failed students may decrease if grace marks help a student reach the passing mark of 40 in Mathematics.
3. Marks cannot exceed 100 because it is the maximum possible mark.

## 5.7 Conclusion

NumPy makes numerical calculations easier and faster through array operations. It helps calculate averages, totals, standard deviations and counts efficiently.

```python
# TASK 5: SPEED UP WITH NUMPY

import csv
import numpy as np

# 1. Read first 10 students from the dataset
students = []

with open("/content/StudentsPerformance.csv", "r", encoding="utf-8-sig") as file:
    reader = csv.DictReader(file)

    for i, row in enumerate(reader):
        if i == 10:
            break

        students.append([
            int(row["math score"]),
            int(row["reading score"]),
            int(row["writing score"])
        ])

# 2. Convert marks into a NumPy array
marks_array = np.array(students)

print("ORIGINAL MARKS ARRAY")
print(marks_array)

print("\nArray Shape:", marks_array.shape)

# 3. Subject-wise average
subject_average = np.mean(marks_array, axis=0)

print("\nSUBJECT-WISE AVERAGE")
print(subject_average)
print("Shape:", subject_average.shape)

# 4. Student-wise total
student_total = np.sum(marks_array, axis=1)

print("\nSTUDENT-WISE TOTAL")
print(student_total)
print("Shape:", student_total.shape)

# 5. Subject-wise standard deviation
subject_std = np.std(marks_array, axis=0)

print("\nSUBJECT-WISE STANDARD DEVIATION")
print(subject_std)
print("Shape:", subject_std.shape)

# 6. Count marks below 40 in each subject
below_40 = marks_array < 40

subject_fail_count = np.sum(below_40, axis=0)

subjects = ["Mathematics", "Reading", "Writing"]

print("\nMARKS BELOW 40 IN EACH SUBJECT")

for i in range(3):
    print(subjects[i], ":", subject_fail_count[i])

print("Shape:", subject_fail_count.shape)

# 7. Find subject with the most marks below 40
most_failed_subject = subjects[np.argmax(subject_fail_count)]

print("\nSUBJECT WITH MOST MARKS BELOW 40:")
print(most_failed_subject)

# 8. Count failed students before grace marks
failed_before = np.any(marks_array < 40, axis=1)

print("\nFAILED STUDENTS BEFORE GRACE MARKS:")
print(np.sum(failed_before))

# 9. Add 5 grace marks to Mathematics
grace_array = marks_array.copy()

grace_array[:, 0] = np.minimum(
    grace_array[:, 0] + 5, 100
)

print("\nARRAY AFTER GRACE MARKS")
print(grace_array)

# 10. Count failed students after grace marks
failed_after = np.any(grace_array < 40, axis=1)

print("\nFAILED STUDENTS AFTER GRACE MARKS:")
print(np.sum(failed_after))

# 11. Compare the results
print("\nGRACE MARKS COMPARISON")

print("Failed before grace marks:", np.sum(failed_before))
print("Failed after grace marks:", np.sum(failed_after))

if np.sum(failed_after) < np.sum(failed_before):
    print("The number of failed students decreased.")
elif np.sum(failed_after) == np.sum(failed_before):
    print("The number of failed students remained the same.")
else:
    print("The number of failed students increased.")
```

I created Student Performance 2 by adding an attendance percentage column to the original dataset. This allows me to analyse attendance and its relationship with students' academic performance. The attendance values are simulated for practice purposes.

```python
# TASK 6: ANALYSE THE FULL CLASS USING PANDAS

# 1. Import library and load dataset

import pandas as pd

df = pd.read_csv("/content/Student_Performance_2.csv")

subjects = ["math score", "reading score", "writing score"]

print("DATASET OVERVIEW")
print("Shape:", df.shape)
print("\nColumn Names:", df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Records:", df.duplicated().sum())

print("\nFirst Five Records:")
display(df.head())


# 2. Data Cleaning

# Fill missing numerical values using median
for col in df.select_dtypes(include="number").columns:
    df[col] = df[col].fillna(df[col].median())

# Fill missing categorical values using mode
for col in df.select_dtypes(include="object").columns:
    if df[col].isnull().any():
        df[col] = df[col].fillna(df[col].mode()[0])

# Remove duplicate records
df = df.drop_duplicates().copy()

# Generate record IDs after cleaning
df.insert(0, "Record_ID", range(1, len(df) + 1))

print("\nDATA AFTER CLEANING")
print("Shape:", df.shape)
print("Missing Values:", df.isnull().sum().sum())
print("Duplicate Records:", df.duplicated().sum())


# 3. Define Grading Functions

def get_grade(marks):
    if marks >= 90:
        return "A"
    elif marks >= 75:
        return "B"
    elif marks >= 60:
        return "C"
    elif marks >= 40:
        return "D"
    else:
        return "F"


def check_result(marks):
    if any(mark < 40 for mark in marks):
        return "Fail"
    else:
        return "Pass"


def calculate_average(marks):
    return sum(marks) / len(marks)


# 4. Calculate Total, Percentage, Grade and Result

df["Total"] = df[subjects].sum(axis=1)

df["Percentage"] = df["Total"] / 3

df["Overall Grade"] = df["Percentage"].apply(get_grade)

df["Result"] = df[subjects].apply(
    lambda row: check_result(row.tolist()), axis=1
)

print("\nSTUDENT PERFORMANCE")
display(df.head(10))


# 5. Average Marks in Each Subject

print("\nAVERAGE MARKS IN EACH SUBJECT")

subject_averages = df[subjects].mean()

print(subject_averages.round(2))


# 6. Identify Failed Students and Failed Subjects

failed_students = df[df["Result"] == "Fail"].copy()

failed_students["Failed Subjects"] = failed_students[subjects].apply(
    lambda row: ", ".join(
        [col for col in subjects if row[col] < 40]
    ),
    axis=1
)

print("\nTOTAL FAILED STUDENTS:", len(failed_students))

print("\nFAILED STUDENTS AND THEIR SUBJECTS")

display(
    failed_students[
        ["Record_ID"] + subjects + ["Failed Subjects"]
    ]
)

print("\nSTUDENTS BELOW 40 IN EACH SUBJECT")

print(df[subjects].lt(40).sum())


# 7. Grade Distribution

grade_counts = df["Overall Grade"].value_counts()

grade_counts = grade_counts.reindex(
    ["A", "B", "C", "D", "F"], fill_value=0
)

print("\nGRADE DISTRIBUTION")
print(grade_counts)


# 8. Identify Top Five Students

top_five = df.nlargest(5, "Percentage")

print("\nTOP FIVE STUDENT RECORDS")

display(
    top_five[
        ["Record_ID"] + subjects +
        ["Total", "Percentage", "Overall Grade"]
    ]
)


# 9. Identify At-Risk Students

at_risk = df[
    (df["attendance_percentage"] < 75) &
    (df["Percentage"] < 50)
]

print("\nTOTAL AT-RISK STUDENTS:", len(at_risk))

print("\nAT-RISK STUDENTS")

display(
    at_risk[
        ["Record_ID", "attendance_percentage",
         "Percentage", "Overall Grade", "Result"]
    ]
)


# 10. Average Attendance by Grade

attendance_by_grade = (
    df.groupby("Overall Grade")["attendance_percentage"]
    .mean()
    .reindex(["A", "B", "C", "D", "F"])
)

print("\nAVERAGE ATTENDANCE BY GRADE")

print(attendance_by_grade.round(2))


# 11. Correlation Between Attendance and Academic Performance

correlation = df["attendance_percentage"].corr(
    df["Percentage"]
)

print("\nATTENDANCE AND PERCENTAGE CORRELATION")

print(round(correlation, 4))


# 12. Compare Task 4 and Task 6 Using the Same First 10 Records

first_10 = df.head(10)

# Plain Python calculation
python_results = []

for _, row in first_10.iterrows():

    marks = [row[sub] for sub in subjects]

    total = sum(marks)
    percentage = calculate_average(marks)
    grade = get_grade(percentage)
    result = check_result(marks)

    python_results.append(
        [total, percentage, grade, result]
    )


# Pandas calculation
pandas_results = first_10[
    ["Total", "Percentage", "Overall Grade", "Result"]
].values.tolist()


# Compare both methods
comparison = []

for i in range(10):

    comparison.append({
        "Record_ID": first_10.iloc[i]["Record_ID"],
        "Python Total": python_results[i][0],
        "Pandas Total": pandas_results[i][0],
        "Python Percentage": python_results[i][1],
        "Pandas Percentage": pandas_results[i][1],
        "Python Grade": python_results[i][2],
        "Pandas Grade": pandas_results[i][2],
        "Python Result": python_results[i][3],
        "Pandas Result": pandas_results[i][3]
    })


comparison_df = pd.DataFrame(comparison)

print("\nTASK 4 AND TASK 6 COMPARISON")

display(comparison_df)


# Final comparison
same_results = all(
    python_results[i] == pandas_results[i]
    for i in range(10)
)

print("\nDO BOTH METHODS GIVE THE SAME RESULTS?", same_results)
```
